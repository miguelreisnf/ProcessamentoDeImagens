```python
img_rgb = cv2.imread('optimus_prime.jpg', cv2.IMREAD_COLOR_RGB)
plt.imshow(img_rgb)
img_gray = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2GRAY)
plt.imshow(img_gray, cmap='gray')
```

```python
def calcHistGrayScale(img):
  hist_arr = np.zeros(256, dtype=int)

  for px in img:
    hist_arr[px] += 1

  plt.bar(range(256), hist_arr)
  plt.xlabel('Intensidade de Pixel')
  plt.ylabel('Frequencia')
  plt.title('Histograma de Imagem')
  plt.show()
```

```python
calcHistGrayScale(img_gray)
```
